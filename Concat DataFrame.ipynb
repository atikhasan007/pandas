# pandas Concat

```python
import pandas as pd

Bd_weather = pd.DataFrame({
    'city' : ['Dhaka','Rajshahi','khulna'],
    'temperature':[32,45,30],
    'humidity':[80,60,78]
})

Bd_weather
```

```python

Us_weather = pd.DataFrame({
    'city' : ['New york','chicago','orlando'],
    'temperature':[21,14,35],
    'humidity':[68,65,75]
})

Us_weather
```

```python
pd.concat([Bd_weather, Us_weather])
```

```python
pd.concat([Bd_weather, Us_weather],ignore_index=True)
```

# keys

```python
df = pd.concat([Bd_weather, Us_weather], keys=["Bangladesh","us"])
df
```

```python
df.loc["Bangladesh"]
```

```python
df.loc["us"]
```

```python
temperature_df = pd.DataFrame({
   'city' : ['Dhaka','Rajshahi','khulna'],
    'temperature':[32,45,30],
})

temperature_df
```

```python
windspeed_df =pd.DataFrame(  {
     'city' : ['Dhaka','Rajshahi','khulna'],
    'windspeed':[7,12,9],
})
windspeed_df
```

```python
df = pd.concat([
temperature_df, windspeed_df])
df
```

```python
df = pd.concat([
temperature_df, windspeed_df],axis=1)
df
```
