# Merge DataFrame

```python
import pandas as pd

df1 = pd.DataFrame({
    'city' : ["new york","chicago","orlando"],
    'temperature':[21, 14, 35]
    
})


df1
```

```python
df2 = pd.DataFrame({
    'city' : ["chicago","new york","orlando"],
    'humidity':[65,68,75]
    
})

df2
```

```python
pd.merge(df1,df2,on="city")
```

```python


df1 = pd.DataFrame({
    'city' : ["new york","chicago","orlando","bultimol"],
    'temperature':[21, 14, 35,45]
    
})


df1
```

```python
df2 = pd.DataFrame({
    'city' : ["chicago","new york","sunfrisco"],
    'humidity':[65,68,75]
    
})

df2
```

```python
df3 = pd.merge(df1,df2, on="city",how="outer")
df3
```

```python
df4 = pd.merge(df1,df2, on="city",how="left")
df4
```

```python
df4 = pd.merge(df1,df2, on="city",how="right")
df4
```

```python
df4 = pd.merge(df1,df2, on="city",how="inner")
df4
```

# use indicator flag in join

```python
df5 = pd.merge(df1,df2, on="city",how="outer" , indicator= True)

df5
```

```python
df6 = pd.DataFrame({
    'city':['new york','chicago','orlando','baltimore'],
    'temperature':[21,14,35,38],
    'humidity':[65,68,71,75]
})

df6
```

```python
pd.merge(df6,df7,on='city')
```
