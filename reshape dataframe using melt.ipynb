# melt is used to transform or reshape data

```python
import pandas as pd
df = pd.DataFrame([
    {'day':'Monday','chicago':'32','chennal':75,'berlin':41},
     {'day':'Tuesday','chicago':'32','chennal':75,'berlin':41},
     {'day':'Wednesday','chicago':'32','chennal':75,'berlin':41},
     {'day':'Thursday','chicago':'32','chennal':75,'berlin':41},
     {'day':'Friday','chicago':'32','chennal':75,'berlin':41},
     {'day':'Saturday','chicago':'32','chennal':75,'berlin':41},
     {'day':'Sunday','chicago':'32','chennal':75,'berlin':41},
])

df
```

```python
df1 = pd.melt(df, id_vars=["day"])
df1
```

```python
df1 = pd.melt(df, id_vars=["day"])
df1[df1['variable']=='chicago']
```

```python
df1 = pd.melt(df, id_vars=["day"])
df1[df1['variable']=='berlin']
```

```python
df1 = pd.melt(df, id_vars=["day"],var_name='city',value_name='temperature')
df1
```
