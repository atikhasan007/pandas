```python
import pandas as pd
from sqlalchemy import create_engine
from pandas.io import sql
```

```python
engine = create_engine("mysql+pymysql://{user}:{pw}@localhost:3306/{db}"
                      .format(user="root",
                             pw="Root%401234",
                              db="application"
                             )
                      )
```

```python
try:
    connceton = engine.connect()
    print("successfull")
except Exception as e:
    print("error message",e)
```

```python
df = pd.read_sql_table('customers', engine, columns=['name','phone_number'])
```

```python
query='''
SELECT
customers.name,
customers.phone_number,

orders.name AS order_name,
orders.amount

from customers
inner join orders
    ON customers.id = orders.customer_id;

'''
df = pd.read_sql_query(query, engine)
df
```

```python
data = pd.DataFrame([
    {'Customer Name':'bulbul','Customer Phone':'232454545'},
    {'Customer Name':'shorif','Customer Phone':'232444545'},
    {'Customer Name':'roton','Customer Phone':'232455545'},
    {'Customer Name':'sohel','Customer Phone':'2222454545'},
    {'Customer Name':'akbor','Customer Phone':'23245333545'},
    {'Customer Name':'rakib','Customer Phone':'238884545'},
])

data
```

```python
data.rename(columns={
    'Customer Name':'name',
    'Customer Phone':'phone_number'
}, inplace=True)

data #	 inplace=True এর মানে হলো, এই পরিবর্তনটা মূল data DataFrame-এ সরাসরি হয়ে যাবে — আলাদা করে কিছু রিটার্ন করার দরকার নেই
```

```python
data.to_sql(
name='customers',
    con=engine,
    index=False,
    if_exists='append'
    
)
```
