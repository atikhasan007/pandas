```python
import pandas as pd

# MultiIndex columns definition
columns = pd.MultiIndex.from_tuples([
    ('Price Ratios', 'Price', 'Facebook'),
    ('Price Ratios', 'Price', 'Google'),
    ('Price Ratios', 'Price', 'Microsoft'),
    ('Price Ratios', 'Price to earnings ratio (P/E)', 'Facebook'),
    ('Price Ratios', 'Price to earnings ratio (P/E)', 'Google'),
    ('Price Ratios', 'Price to earnings ratio (P/E)', 'Microsoft'),
    ('Income Statement', 'Net Sales', 'Facebook'),
    ('Income Statement', 'Net Sales', 'Google'),
    ('Income Statement', 'Net Sales', 'Microsoft'),
    ('Income Statement', 'Net Profit', 'Facebook'),
    ('Income Statement', 'Net Profit', 'Google'),
    ('Income Statement', 'Net Profit', 'Microsoft'),
])

# Row index
index = ['Q1 2016', 'Q2 2016', 'Q3 2016', 'Q4 2016', 'Q1 2017']

# Data values
data = [
    [155, 955, 66, 37.10, 32.0, 30.31, 2.6, 20, 18.70, 0.80, 5.43, 4.56],
    [150, 987, 69, 36.98, 31.3, 30.56, 3.1, 22, 21.30, 0.97, 5.89, 5.10],
    [153, 963, 62, 36.78, 31.7, 30.46, 4.3, 24, 21.45, 1.20, 6.10, 5.43],
    [155, 1000, 61, 36.11, 31.2, 30.11, 6.7, 26, 21.88, 1.67, 6.50, 5.89],
    [156, 1012, 66, 37.07, 30.0, 31.00, 8.1, 31, 22.34, 2.03, 6.40, 6.09],
]

# Create DataFrame
df = pd.DataFrame(data, index=index, columns=columns)

# Display
print(df)
```
