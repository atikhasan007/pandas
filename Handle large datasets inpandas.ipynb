# Technique 2 : shrink numerical columns with smaller dtypes

```python

import pandas as pd
import numpy as np

df = pd.read_csv('/home/catpc/Downloads/weather_data_3city.csv')
df
```

```python
df['temperature'].memory_usage(index=False,deep=True)
```

```python
df['temperature'].max()
```

```python
df['temperature'].min()
```

```python
df = pd.read_csv('/home/catpc/Downloads/weather_data_3city.csv',
                dtype={'temperature':'int8'}
                )


df['temperature'].memory_usage(index=False, deep=True)
```
